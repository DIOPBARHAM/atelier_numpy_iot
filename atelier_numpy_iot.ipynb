## Partie 1 : Listes Python vs tableaux NumPy

### 1 . Créer une liste Python stockant ces températures (24.5, 25.1, 26.3, 27.0 et 25.8)

```python
# Création d'une liste Python 
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]

print("Liste Python :", temperatures_liste)
```

### 2 . Créer un tableau NumPy équivalent stockant ces températures

```python
import numpy as np

# Conversion de la liste en tableau NumPy 
temperatures_numpy = np.array([24.5, 25.1, 26.3, 27.0, 25.8])

print("Tableau NumPy :", temperatures_numpy)
```

### 3 . Effectuer une multiplication par 2 sur ces deux collections

```python
# Multiplication par 2 de la liste Python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
liste_multipliee = temperatures_liste * 2

# Multiplication par 2 du tableau NumPy
temperatures_numpy = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
numpy_multiplie = temperatures_numpy * 2

# Affichage des deux résultats
print("Résultat liste Python :", liste_multipliee)
print("Résultat tableau NumPy :", numpy_multiplie)
```

### 4 . Quelle différence observe-t-on ?

Différence : La liste Python a été dupliquée (répétition des éléments),alors que le tableau NumPy a subi une multiplication vectorielle de chaque élément.

## Partie 2 : Création des données

### 1 . Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro


```python
# Initialisation d'un tableau de 10 zéros (floats par défaut)
import numpy as np
capteurs_zeros = np.zeros(10)

# Affichage du tableau
print("Capteurs à zéro :", capteurs_zeros)
```

### 2 . Créer un tableau contenant 10 valeurs égales à 1


```python
import numpy as np
# Initialisation d'un tableau de 10 uns
capteurs_uns = np.ones(10)

print("Capteurs à un :", capteurs_uns)
```

### 3 . Créer un tableau de 10 valeurs initialisées à 25

```python
import numpy as np
# Initialisation d'un tableau de 10 éléments contenant chacun la valeur 25
capteurs_25 = np.full(10, 25.0)
print("Capteurs à 25 :", capteurs_25)
```

### 4 . Créer une matrice identité de taille 5 × 5

```python
# Création d'une matrice identité 5x5
import numpy as np
matrice_identite = np.eye(5)
print("Matrice identité 5x5 :\n", matrice_identite)
```

###  5 . Créer les 24 heures (de 0 à 23) de mesure d’une journée

```python
# Génération d'un tableau allant de 0 à 23 inclus 
import numpy as np
heures_journee = np.arange(0, 24)

print("Heures de la journée :", heures_journee)
```

### 6 . Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C

```python
import numpy as np
temperatures_linspace = np.linspace(20, 30, 10)
print("Températures réparties :", temperatures_linspace)
```

### 7 . Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles

```python
import numpy as np
np.random.seed(42)

# Génération de 100 températures aléatoires uniformes entre 15 °C et 35 °C
mesures_aleatoires = np.random.uniform(15, 35, 100)

print("10 premières mesures de température aléatoires :\n", mesures_aleatoires[:10])
```
