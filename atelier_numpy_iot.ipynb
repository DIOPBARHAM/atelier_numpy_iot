## Partie 1 : Listes Python vs tableaux NumPy

### 1 . Créer une liste Python stockant ces températures (24.5, 25.1, 26.3, 27.0 et 25.8)

```python
# Création d'une liste Python 
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]

print("Liste Python :", temperatures_liste)
```

### 2 . Créer un tableau NumPy équivalent stockant ces températures

```python
import numpy as np

# Conversion de la liste en tableau NumPy 
temperatures_numpy = np.array([24.5, 25.1, 26.3, 27.0, 25.8])

print("Tableau NumPy :", temperatures_numpy)
```

### 3 . Effectuer une multiplication par 2 sur ces deux collections

```python
# Multiplication par 2 de la liste Python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
liste_multipliee = temperatures_liste * 2

# Multiplication par 2 du tableau NumPy
temperatures_numpy = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
numpy_multiplie = temperatures_numpy * 2

# Affichage des deux résultats
print("Résultat liste Python :", liste_multipliee)
print("Résultat tableau NumPy :", numpy_multiplie)
```

### 4 . Quelle différence observe-t-on ?

Différence : La liste Python a été dupliquée (répétition des éléments),alors que le tableau NumPy a subi une multiplication vectorielle de chaque élément.

## Partie 5 : Filtrage booléen

### 1. Afficher les températures supérieures à 30 °C

```python
import numpy as np
np.random.seed(42)
temperatures_larges = np.random.uniform(15, 40, 24)

# Étape 1 : Créer la condition logique, Étape 2 : L'appliquer en indexation
temperatures_chaudes = temperatures_larges[temperatures_larges > 30]

print("Mesures de température larges :\n", temperatures_larges)
print("Températures supérieures à 30 °C :", temperatures_chaudes)
```

### 2 . Afficher les températures « normales » càd comprises entre 20 °C et 30 °C

```python
import numpy as np
np.random.seed(10)
temperatures_larges = np.random.uniform(15, 40, 24)
# Combinaison des filtres >= 20 et <= 30
temperatures_normales = temperatures_larges[(temperatures_larges >= 20) & (temperatures_larges <= 30)]

print("Températures normales (20 à 30 °C inclus) :\n", temperatures_normales)
```

### 3 . Supposons qu'une température supérieure à 35 °C soit considérée comme anormale. Déterminer anomalies, un tableau de booléen précisant les anomalies

```python
import numpy as np
np.random.seed(10)
temperatures_larges = np.random.uniform(15, 40, 24)
# Génération du masque de booléens pour les valeurs strictement supérieures à 35
anomalies = temperatures_larges > 35

print("Masque logique d'anomalies (>35 °C) :\n", anomalies)
```

### 4 . Déterminer le nombre de températures anormales à l’aide de la variable anomalies

```python
import numpy as np
np.random.seed(10)
temperatures_larges = np.random.uniform(15, 40, 24)
# Génération du masque de booléens pour les valeurs strictement supérieures à 35
anomalies = temperatures_larges > 35
# Somme du tableau de booléens pour compter les True
nombre_anomalies = np.sum(anomalies)

print("Nombre total de températures anormales :", nombre_anomalies)
```
